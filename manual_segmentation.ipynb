```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import json
import cv2
import numpy as np

mpl.rc('figure',  figsize=(20, 10))
mpl.rc('image', cmap='gray')

# Read in the JSON file
with open('labels.json', 'r') as f:
    segmentation_data = json.load(f)

# Extract the segmentation information
segmentation_mask = np.array(segmentation_data['annotations'])


# Read in the original image
original_image = cv2.imread('./bubble_raw.png')

# Plot the polygons on the image
for object in segmentation_mask:
    points = np.array(object["segmentation"], np.int32)
    points = points.reshape((-1, 1, 2))
    cv2.polylines(original_image, [points], True, (0, 0, 255), 2)


plt.imshow(original_image)
plt.xticks([]), plt.yticks([])  # to hide tick values on X and Y axis
plt.show()

print(len(segmentation_mask))
```
