```python
def rotation(image):
    height,width = image.shape[:2]
    rotation_matrix = cv2.getRotationMatrix2D((width/2,height/2),90,1)
    rotated_image = cv2.warpAffine(image,rotation_matrix,(width,height))
    return rotated_image
```

```python
def sharpening(image):
    kernel = np.array([[0, -1, 0],
                   [-1, 5,-1],
                   [0, -1, 0]])
    image_sharp = cv2.filter2D(src=image, ddepth=-1, kernel=kernel)
    return image_sharp
```

```python
def GaussianBlur(image):
    kernelSizes = [(3, 3), (9, 9), (15, 15)]
    for (kX, kY) in kernelSizes:
        blurred = cv2.GaussianBlur(image, (kX, kY), 0)
    return blurred
```

```python
#salt and pepper noise
def add_noise(img):
    image = img.copy()
    row , col, z = image.shape
    # Randomly pick some pixels in the
    # image for coloring them white
    number_of_pixels = 10000
    for i in range(number_of_pixels):
        y_coord=random.randint(0, row - 1)
        x_coord=random.randint(0, col - 1)
        image[y_coord][x_coord] = 255

    # Randomly pick some pixels in
    # the image for coloring them black
    number_of_pixels = 10000
    for i in range(number_of_pixels):
        y_coord=random.randint(0, row - 1)
        x_coord=random.randint(0, col - 1)
        image[y_coord][x_coord] = 0   
    return image
```

```python
def random_crop(img, scale=0.5):
    
    # Crop image
    height, width = int(img.shape[0]*scale), int(img.shape[1]*scale)
    x = random.randint(0, img.shape[1] - int(width))
    y = random.randint(0, img.shape[0] - int(height))
    cropped = img[y:y+height, x:x+width]
    resized = cv2.resize(cropped, (img.shape[1], img.shape[0]))
    
    return resized
```

```python
def extractImages(pathIn,pathOut):
    count = 0
    vidcap = cv2.VideoCapture(pathIn)
    print(pathOut)
    vid= pathIn.split("/")[6].split(".")[0]
    success,image = vidcap.read()
    success = True
    while success:
        image_augmented = []
        vidcap.set(cv2.CAP_PROP_POS_MSEC,(count*500)) # added this line
        success,image = vidcap.read()
        #save original images 
        #cv2.imwrite(pathOut + "/" + vid+"- imageOrigin"+str(count)+".jpg",image)
        if(success != False):
            image_rotation = rotation(image)
            image_noise = add_noise(image)            
            image_sharpening = sharpening(image)
            image_gaussianblur = GaussianBlur(image)
            image_randomCrop = random_crop(image)
            image_augmented = image_augmented + [image_rotation,image_noise,image_sharpening,image_gaussianblur,image_randomCrop]
            
            if(os.path.exists(pathOut)==False):
                os.mkdir(pathOut)
                
            c = 0
            for i in image_augmented:
                pathToSave= pathOut + "/" + vid+"-"+str(count)+"-"+str(c)+".jpg"
                cv2.imwrite(pathToSave,i) # save frame as JPEG file
                print("frame "+vid+"-"+str(count)+"-"+str(c)+".jpg",success) 
                c=c+1
        count = count+1 
```

```python
for video in videos :
    extractImages(path_in+video, path_out+video.replace("mp4",""))
```
